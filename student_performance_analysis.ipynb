- Visualization 1: Class Attendance Percentage vs GPA/CGPA

Objective

To explore how students' reported class attendance percentages are distributed across different GPA/CGPA categories.

Analysis

This visualization compares attendance categories with GPA/CGPA categories. The number of students in each combination helps show how the responses are distributed.

The analysis focuses on:

Which attendance categories contain the largest numbers of respondents.
How GPA/CGPA categories are distributed within each attendance category.
Whether higher attendance categories appear to contain different GPA/CGPA distributions from lower attendance categories.
Whether any noticeable patterns or exceptions appear in the dataset.
Interpretation

Use the chart to describe the patterns actually visible in the collected data. If higher attendance categories contain more students from certain GPA/CGPA categories, report that as an observed association within this sample.

The chart alone cannot establish that attendance causes better academic performance. Other factors, including study habits, course difficulty, and individual circumstances, may also contribute.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv("E:\8th semester\data-science-journey\project1\Student Performance Data Analysis (Responses) - Form Responses 1.csv")

att_gpa =pd.crosstab(df["Approximately what percentage of your classes do you attend?"],df["What was your approximate GPA/CGPA in your most recently completed semester?"])
att_gpa_pr =pd.crosstab(df["Approximately what percentage of your classes do you attend?"],df["What was your approximate GPA/CGPA in your most recently completed semester?"],normalize="index") * 100

print(att_gpa)
print(att_gpa_pr.round(1))
att_gpa.plot(
    kind="bar",
    # stacked=True,
    # figsize=(9, 5),
    title="Attendance vs GPA"
)
plt.legend()
plt.show()
```

- Visualization 2: Academic Confidence vs GPA/CGPA

Objective

To explore the relationship between students' self-reported academic confidence and their reported GPA/CGPA categories.

Analysis

This visualization compares students' confidence responses with their GPA/CGPA categories.

The analysis considers:

The distribution of GPA/CGPA categories across confidence levels.
Whether students reporting higher confidence are concentrated in particular GPA/CGPA categories.
Whether students reporting lower confidence appear in multiple GPA/CGPA categories.
Whether the observed responses show any unexpected differences between confidence and reported performance.
Interpretation

Academic confidence represents students' perceptions of their own performance, while GPA/CGPA represents their reported academic results.

The chart can help explore whether these two measures appear to align in the collected sample. However, confidence and GPA/CGPA are different measures, and the visualization does not prove that one causes the other.

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("E:\8th semester\data-science-journey\project1\Student Performance Data Analysis (Responses) - Form Responses 1.csv")

conf_cg = pd.crosstab(df["How confident are you about your academic performance this semester?"],df["What was your approximate GPA/CGPA in your most recently completed semester?"])
conf_cg.plot(kind="bar")
plt.legend(title="CGPA")

# plt.hist(df["How confident are you about your academic performance this semester?"], bins=5, color='orange')
plt.xlabel("Confidence Label")
plt.ylabel("Number of students")
plt.show()
```

- Visualization 4: Daily Study Hours vs GPA/CGPA

Objective

To investigate how students' reported daily study hours outside classes are distributed across GPA/CGPA categories.

Analysis

This visualization compares daily study-hour categories with reported GPA/CGPA categories.

The analysis considers:

Which daily study-hour categories are most common.
How GPA/CGPA categories are distributed among students with different study durations.
Whether particular study-hour categories appear more frequently alongside certain GPA/CGPA categories.
Whether the pattern appears consistent or contains notable exceptions.
Interpretation

Use the chart to describe the relationship between reported study duration and GPA/CGPA within this sample.

Study duration alone does not determine academic success. Study quality, consistency, learning methods, course difficulty, and other circumstances may also influence performance.

Because study hours are collected as survey categories, avoid treating them as exact numerical measurements unless the original responses provide numerical values.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv("E:\8th semester\data-science-journey\project1\Student Performance Data Analysis (Responses) - Form Responses 1.csv")


# Count students in each study-hour category
study_counts = df["On average, how many hours do you study outside classes per day?"].value_counts()

print(study_counts)
study_percentage = df["On average, how many hours do you study outside classes per day?"].value_counts(normalize=True) * 100

print(study_percentage.round(2))


study_counts.plot(
    kind="bar",
    title="Distribution of Daily Study Hours",
    xlabel="Study-hour category",
    ylabel="Number of students"
)
plt.legend()

# plt.tight_layout()  # Automatically fixes spacing so labels don't get cut off or overlap
plt.show()
```

## Preliminary Academic Risk Review

Objective

To identify GPA/CGPA categories that may warrant further supportive review.

Method

A simple rule-based classification is used to flag selected GPA/CGPA categories:

2.00–2.49: High concern — review support needs.
2.50–2.99: Needs attention — review support needs.
Other GPA/CGPA categories: No GPA-based flag.

Missing GPA/CGPA values should be identified separately so that incomplete records are not mistakenly treated as evidence of good academic performance.

Interpretation

This classification is a preliminary screening rule based on reported GPA/CGPA categories. It is not a machine-learning model, a validated prediction system, or a definitive assessment of any student's academic situation.

The thresholds should be checked against the applicable institutional grading policy. A student without a GPA-based flag may still face academic difficulties and may need support.

The purpose of this analysis is to demonstrate how survey data can be used to organize records for possible supportive follow-up, not to label or judge individual students.

Unmeasured factors: Personal circumstances, teaching methods, course difficulty, and other variables may affect academic performance but may not be captured adequately by the survey.
